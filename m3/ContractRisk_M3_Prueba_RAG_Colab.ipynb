```python
from google.colab import files
from pathlib import Path
import zipfile, os
uploaded = files.upload()
name = 'contractrisk_m3_smoke_codigo.zip'
assert name in uploaded, f'Sube {name}'
root = Path('/content/contractrisk_m3_smoke')
assert not root.exists(), 'Usa un entorno nuevo para no mezclar corridas.'
with zipfile.ZipFile(name) as z:
    for member in z.namelist():
        assert (root / member).resolve().is_relative_to(root.resolve()), 'Ruta insegura'
    z.extractall(root)
os.chdir(root)
print(root)
```
